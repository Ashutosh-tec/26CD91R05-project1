```python

import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Load your CSV file
df = pd.read_csv("/home/exam/26CD91R05-project1/fig-3/Code frequency.csv")

# Convert dates
df["DateTime"] = pd.to_datetime(df["DateTime"])

# Create figure
fig, ax = plt.subplots(figsize=(16, 6))

# Additions (upward green bars)
ax.bar(
    df["DateTime"],
    df["Additions"],
    width=5,
    color="green",
    label="Additions",
    alpha=0.8
)

# Deletions (downward red bars)
ax.bar(
    df["DateTime"],
    df["Deletions"],
    width=5,
    color="red",
    label="Deletions",
    alpha=0.8
)

# Zero baseline
ax.axhline(0, color="black", linewidth=1)

# X-axis: Display years only
ax.xaxis.set_major_locator(mdates.YearLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))

# Labels and title
ax.set_xlabel("Year")
ax.set_ylabel("Number of Lines Changed")
ax.set_title("Weekly Additions and Deletions")

# Grid and legend
ax.legend()
ax.grid(axis="y", linestyle="--", alpha=0.3)

# Layout
plt.tight_layout()

# Save chart
plt.savefig(
    "fig3.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
