```python

import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("Top committers.csv")

# Create bar chart
plt.figure(figsize=(14, 6))

plt.bar(df["Category"], df["Commits"], color="green", width=0.7)

# Labels and title
plt.xlabel("Committers", fontsize=12)
plt.ylabel("Commits", fontsize=12)
plt.title("Commits by Contributor", fontsize=15)

# Formatting
plt.xticks(rotation=45, ha="right")
plt.yticks(range(0, 6))
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()

# Save image
plt.savefig("fig4.png", dpi=300, bbox_inches="tight")

# Display chart
plt.show()
```
