```python
import pandas as pd
import matplotlib.pyplot as plt

# Data
data = pd.read_csv("/home/exam/26CD91R05-project1/fig-1/Commits over time.csv")

# Create DataFrame
df = pd.DataFrame(data)

# Convert dates to datetime
df["Week of"] = pd.to_datetime(df["Week of"])

# Filter data from 8 August 2026 onward
start_date = "2026-08-08"
df = df[df["Week of"] >= start_date]

# Create bar chart
plt.figure(figsize=(15, 6))

plt.bar(
    df["Week of"],
    df["Commits"],
    width=5,
    color="blue"
)

# Chart formatting
plt.title("Commits Over Time", fontsize=16)
plt.xlabel("Week", fontsize=12)
plt.ylabel("Number of Commits", fontsize=12)

plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()

# Save chart
plt.savefig("fig1.png", dpi=300, bbox_inches="tight")

# Display chart
plt.show()
```
